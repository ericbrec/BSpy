# Point containment in $n$ dimensions

What follows can be found in *Generalized Functions Volume 1:  Properties and Operations*, by I. M. Gel'fand and G. E. Shilov, New York:  Academic Press, 1964.

Theorem:  The delta distribution $\delta(x)$ in $k$ dimensions ($k\ge 3$) satisfies

$$-(k-2)S_n\delta(x-p)=\nabla\cdot\nabla{1\over \|x-p\|^{k-2}}$$

where $S_k$ is the volume of the unit sphere in $\Re^k$.

Theorem:  The delta distribution $\delta(x)$ in $2$ dimensions satisfies

$$-2\pi\delta(x-p)=\nabla\cdot\nabla\log\left({1\over \|x-p\|}\right).$$

Theorem:  The delta distribution $\delta(x)$ in $1$ dimension satisfies

$$2\delta(x-p)=\nabla\cdot\nabla((x-p)_+).$$

Given a region $\Omega$ in $\Re^k$, a point $p$ will be contained in that region if and only if

$$\int_\Omega\delta(x-p)=1.$$

Going through each of the cases in turn.  For $k=1$,

$$\begin{align}\int_\Omega\delta(x-p)&={1\over2}\int_\Omega\nabla\cdot\nabla((x-p)_+)\cr
&={1\over2}\int_{\partial\Omega}\nabla((x-p)_+)\cdot n\cr
&={1\over2}\int_{\partial\Omega}(x-p)_+^0\cdot n\cr
&={1\over2}\int_{\partial\Omega}{(x-p)\cdot n\over\|x-p\|}.\cr\end{align}$$

For $k=2$,

$$\begin{align}\int_\Omega\delta(x-p)&=-{1\over 2\pi}\int_\Omega\nabla\cdot\nabla\log({1\over\|x-p\|})\cr
&=-{1\over 2\pi}\int_{\partial\Omega}\nabla\log({1\over\|x-p\|})\cdot n\cr
&={1\over 2\pi}\int_{\partial\Omega}{(x-p)\cdot n\over\|x-p\|^2}\cr\end{align}$$

For $k\ge3$,

$$\begin{align}\int_\Omega\delta(x-p)&=-{1\over (k-2)S_k}\int_\Omega\nabla\cdot\nabla{1\over\|x-p\|^{k-2}}\cr
&=-{1\over (k-2)S_k}\int_{\partial\Omega}\nabla{1\over\|x-p\|^{k-2}}\cdot n\cr
&={1\over S_k}\int_{\partial\Omega}{(x-p)\cdot n\over\|x-p\|^k}.\cr\end{align}$$

This all suggests to me that the definition of Brouwer index has to be

$$\text{index}_k(p):={1\over S_k}\sum_j\int_{B_{k-1,j}}{(x-p)\cdot n\over\|x-p\|^k}.$$

Of course, one doesn't actually integrate over the $k-1$ dimensional boundary surfaces $B_{k-1,j}$, but over their parameter spaces.  That introduces a change of variables which in turn leads to a measure introduced to the integrand which amounts to a factor of $\|x-p\|^{k-1}$ there which is how this integral transforms into the more familiar one.

Here's an example.  Consider any point $p$ in $\Re^2$.  Suppose there is only one boundary surface, namely the unit circle.  Suppose that the unit circle $C$ is parametrized by

$$C(u)=\begin{pmatrix}\cos 2\pi u\cr \sin 2\pi u\cr\end{pmatrix}.$$

Here's what happens going through the revised definition from start to finish

$$\begin{align}\text{index}_2(p)&={1\over 2\pi}\int_C{(x-p)\cdot n\over\|x-p\|^2}\cr
&={1\over 2\pi}\int_0^1{\begin{pmatrix}\cos 2\pi u - p_1\cr \sin 2\pi u - p_2\cr\end{pmatrix}
\cdot\begin{pmatrix}\cos 2\pi u\cr \sin 2\pi u\cr\end{pmatrix}\over
\begin{pmatrix}\cos 2\pi u - p_1\cr \sin 2\pi u - p_2\cr\end{pmatrix}\cdot\begin{pmatrix}\cos 2\pi u - p_1\cr \sin 2\pi u - p_2\cr\end{pmatrix}}2\pi du\cr
&=\int_0^1{1-p\cdot\begin{pmatrix}\cos 2\pi u\cr \sin 2\pi u\cr\end{pmatrix}\over
1 - 2p\cdot\begin{pmatrix}\cos 2\pi u\cr \sin 2\pi u\cr\end{pmatrix} + \|p\|^2}du\end{align}$$

Here's an example for a line segment:  Suppose that $v_0$ and $v_1$ are points in $\Re^2$.  Then the points on the line segment between them is given by $v(u)=v_0(1-u)+v_1u$.  The contribution of the line segment to the integral around the boundary is given by

$$\begin{align}\text{index}_2(p,v)&=\int_v{(x-p)\cdot n\over\|x-p\|^2}\cr
&=\int_0^1{((v_1-v_0)u+v_0-p)\cdot n\over\|(v_1-v_0)u+v_0-p\|^2}\|v_1-v_0\|du\cr
&=\int_0^1{(v_0-p)\cdot(v_1-v_0)_\perp\over\|(v_1-v_0)u+v_0-p\|^2}du\cr
&=\int_0^1{(v_0-p)\times(v_1-p+p-v_0)\over\|(v_1-v_0)\|^2u^2+2(v_1-v_0)\cdot(v_0-p)u+\|v_0-p\|^2}du\cr
&={2(v_0-p)\times(v_1-p)\over \sqrt{4\|v_0-p\|^2\|v_1-v_0\|^2-4[(v_1-v_0)\cdot(v_0-p)]^2}}\left[\arctan\left({2\|v_1-v_0\|^2u+2(v_1-v_0)\cdot(v_0-p)\over\sqrt{4\|v_0-p\|^2\|v_1-v_0\|^2-4[(v_1-v_0)\cdot(v_0-p)]^2}}\right)\right]_0^1\cr
&=\left[\arctan\left({\|v_1-v_0\|^2u+(v_1-v_0)\cdot(v_0-p)\over(v_0-p)\times(v_1-p)}\right)\right]_0^1\cr
&=\arctan\left({(v_1-v_0)\cdot(v_1-p)\over(v_0-p)\times(v_1-p)}\right)-\arctan\left({(v_1-v_0)\cdot(v_0-p)\over
(v_0-p)\times(v_1-p)}\right)\cr
&=\arctan\left({{\|v_1-v_0\|^2\over(v_0-p)\times(v_1-p)}\over 1+{(v_1-v_0)\cdot(v_1-p)(v_1-v_0)\cdot(v_0-p)\over
[(v_0-p)\times(v_1-p)]^2}}\right)\cr
&=\arctan\left({\|v_1-v_0\|^2(v_0-p)\times(v_1-p)\over [(v_0-p)\times(v_1-p)]^2+(\|v_1-p\|^2-(v_0-p)\cdot(v_1-p))((v_1-p)\cdot(v_0-p)-\|v_0-p\|^2)
}\right)\cr
\end{align}.$$
